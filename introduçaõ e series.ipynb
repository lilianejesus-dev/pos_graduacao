```python
import pandas as pd
```

```python
serie = pd.Series([10,20,30], index=["a", "b", "c"], name= "Idades")
print(serie)
```

```python
idades = pd.Series([25,30,35,40,45, 36,71], name="Idades")
print(idades)
```

```python
print(f"O total de registros é {idades.count()}")
print(f"A media das idades é {idades.mean()}")
print(f"A soma das idades é {idades.sum()}")
print(f"A idades máxima  é {idades.max()}")
print(f"A idade minima é {idades.min()}")
print(f"A mediana das idades é {idades.median()}")
```

```python
print(idades.describe())
```
