```python
import pandas as pd
```

```python
dados = {
    'nome': ['Ana', 'Bruno', 'Carlos', 'Diana'],    
    'idade': [23, 35, 45, 29],
    'salario' : [5000, 7000, 8000, 6000]
}
print(dados)
df = pd.DataFrame(dados)
print(df)
```

```python
dados = [
    ['Ana', 23, 5000],
    ['Bruno', 35, 7000],
    ['Carlos', 45, 8000],
    ['Diana', 29, 6000]
]
df = pd.DataFrame(dados, columns=['nome', 'idade', 'salario'])
print(df)
```
